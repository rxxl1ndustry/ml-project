```python

import numpy as np
import pandas as pd


# ==================================================
# 1-1. CSV 파일 읽기
# ==================================================

# CSV 파일명 지정
file_name = "1.01.house.price.csv"

try:
    # CSV 파일을 읽어 데이터프레임에 저장
    df_house = pd.read_csv(file_name)

except FileNotFoundError:
    # 파일이 존재하지 않을 경우 오류 처리
    print("파일명과 업로드 여부를 확인하세요.")

else:
    # 파일 읽기에 성공한 경우에만 이후 과제 실행

    print("===== 1-1. CSV 파일 읽기 =====")

    # 첫 번째 행 출력
    print("첫 번째 행:")
    print(df_house.iloc[0])

    # 전체 데이터 행 수 출력
    print("전체 데이터 행 수:", len(df_house))

    # 첫 번째 행의 price 자료형 출력
    print("첫 번째 행의 price 자료형:",
          type(df_house.iloc[0]["price"]))


    # ==================================================
    # 1-2. 문자열을 숫자로 변환하기
    # ==================================================

    print("\n===== 1-2. 데이터 정제 =====")

    # 빈 리스트 생성
    houses = []

    # 제외된 행 개수
    excluded_count = 0

    # 각 행을 반복 처리
    for index, row in df_house.iterrows():

        try:
            # 필수 값 누락 여부 확인
            if pd.isna(row["ID"]) or pd.isna(row["size"]) \
                    or pd.isna(row["year"]) or pd.isna(row["price"]):
                raise ValueError("필수 값 누락")

            # 자료형 변환
            house = {
                "ID": int(row["ID"]),
                "size": int(row["size"]),
                "year": int(row["year"]),
                "price": float(row["price"])
            }

            # 정상 데이터 저장
            houses.append(house)

        except (ValueError, TypeError, KeyError) as e:
            print(f"{index}번째 행 제외 사유: {e}")
            excluded_count += 1

    # 정상 저장 및 제외 행 개수 출력
    print("정상 저장 행 수:", len(houses))
    print("제외 행 수:", excluded_count)


    # ==================================================
    # 1-3. 조건에 맞는 주택 찾기
    # ==================================================

    print("\n===== 1-3. 주택 데이터 필터링 =====")

    # 원본 houses를 유지하고 복사본 생성
    houses_test = houses.copy()

    # 테스트 데이터 추가
    houses_test.extend([
        {"ID": 101, "year": 5, "price": 3},
        {"ID": 102, "size": 35, "year": 5, "price": None},
        {"ID": 103, "size": "오류", "year": 5, "price": 3},
        {"ID": 104, "size": 35, "year": 5, "price": 3}
    ])

    # 조건에 맞는 주택을 저장할 리스트
    selected = []

    # 모든 주택 데이터 반복
    for house in houses_test:

        # size 또는 price 키가 없는 경우 제외
        if "size" not in house or "price" not in house:
            print(f"ID {house.get('ID')} 제외: 필수 키 누락")
            continue

        # None 또는 결측값 확인
        if pd.isna(house["size"]) or pd.isna(house["price"]):
            print(f"ID {house.get('ID')} 제외: 값 누락")
            continue

        try:
            # 숫자로 변환
            size = float(house["size"])
            price = float(house["price"])

        except (ValueError, TypeError) as e:
            print(f"ID {house.get('ID')} 제외: 숫자 변환 오류 ({e})")
            continue

        # 크기가 0 이하이거나 가격이 음수인 경우 제외
        if size <= 0 or price < 0:
            print(f"ID {house.get('ID')} 제외: 유효하지 않은 크기 또는 가격")
            continue

        # 크기 30 이상, 가격 4 이하인 주택 선택
        if size >= 30 and price <= 4:
            selected.append(house)

    # 선택된 주택 ID 하나씩 출력
    print("\n선택된 주택 ID:")
    for house in selected:
        print(house["ID"])

    # 선택된 주택 전체 건수 출력
    print("선택된 주택 수:", len(selected))


    # ==================================================
    # 1-4. 함수와 예외 처리
    # ==================================================

    print("\n===== 1-4. 크기당 가격 계산 =====")

    # 가격을 크기로 나누는 함수
    def unit_price(price, size):

        try:
            # 숫자 또는 숫자 형태의 문자열을 숫자로 변환
            price = float(price)
            size = float(size)

            # 크기가 0 이하인 경우
            if size <= 0:
                print("크기 오류: 크기는 0보다 커야 합니다.")
                return None

            # 가격이 음수인 경우
            if price < 0:
                print("가격 오류: 가격은 음수일 수 없습니다.")
                return None

            # 정상 입력이면 가격을 크기로 나눈 값 반환
            return price / size

        except ValueError:
            print("값 오류: 숫자로 변환할 수 없는 문자열입니다.")
            return None

        except TypeError:
            print("자료형 오류: 숫자 또는 숫자 형태의 문자열을 입력하세요.")
            return None


    # 전체 houses에 함수 적용
    print("전체 주택의 크기당 가격:")

    for house in houses:
        result = unit_price(house["price"], house["size"])

        print(
            f"주택 ID: {house['ID']}, "
            f"크기당 가격: {result}"
        )


    # 함수 호출 및 예외 처리 테스트
    print("\n함수 테스트 결과:")

    print("unit_price(4, 40):", unit_price(4, 40))

    print('unit_price("4", "40"):', unit_price("4", "40"))

    print("unit_price(4, 0):", unit_price(4, 0))

    print("unit_price(-4, 40):", unit_price(-4, 40))

    print('unit_price("오류", 40):', unit_price("오류", 40))

    print("unit_price(None, 40):", unit_price(None, 40))

    print("unit_price(6, 30):", unit_price(6, 30))


    # ==================================================
    # 1-5. 목적에 맞는 자료구조 선택하기
    # ==================================================

    print("\n===== 1-5. 자료구조 선택 =====")

    # ① 리스트(list)
    # 원래 순서를 유지하고 중복을 허용해야 하므로 리스트 선택
    # 집합(set)은 중복을 제거하고 원래 순서를 보장하지 않음

    house_ids = [house["ID"] for house in houses]

    print("전체 주택 ID:", house_ids)


    # ② 집합(set)
    # year 값을 중복 없이 저장해야 하므로 집합 선택
    # 리스트(list)는 중복된 값을 그대로 저장함

    unique_years = {house["year"] for house in houses}

    print("중복 없는 year:", unique_years)


    # ③ 딕셔너리(dict)
    # 주택 ID를 키, 가격을 값으로 저장하여 ID로 가격을 조회
    # 리스트(list)는 ID에 해당하는 가격을 찾으려면 별도 탐색이 필요함

    house_prices = {
        house["ID"]: house["price"]
        for house in houses
    }

    if 3 in house_prices:
        print("ID 3의 가격:", house_prices[3])
    else:
        print("ID 3의 주택 데이터가 없습니다.")


    # ④ 튜플(tuple)
    # 최소 크기와 최대 가격을 변경할 수 없도록 튜플 선택
    # 리스트(list)는 원소를 추가, 삭제하거나 교체할 수 있음

    search_conditions = (30, 4)

    print("불변 검색 조건:", search_conditions)
    print("최소 크기:", search_conditions[0])
    print("최대 가격:", search_conditions[1])
```
